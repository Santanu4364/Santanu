```python
num = int(input())
for i in range(1, 11):
    print(f"{num} x {i} = {num * i}")
```

```python
num = int(input())
for i in range(1, 11):
    print('{} X {} = {}'.format(num, i, num * i))
```

```python
num = int(input())
for i in range(1, 11):
    print('{a} X {b} = {c}'.format(a=num, b=i, c=num * i))
```

```python
num = int(input())
for i in range(1, 11):
    print('%d X %d = %d' % (num, i, num*i))
```

```python
pi = 22/7
print(f'value of pi = {pi}')
print('value of pi = {0}' .format(pi))
print('value of pi = %f' % (pi))
```

```python
pi = 22/7
print(f'value of pi = {pi:.2f}')
print('value of pi = {0:.2f}' .format(pi))
print('value of pi = %.2f' % (pi))
```

```python
pi = 22/7
print('%f' % (pi))
```

## Default decimal is 6 for %f
