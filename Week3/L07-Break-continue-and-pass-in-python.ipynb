# Break
#### It will stop iteration in a specific charachter which will be specified.
#### In the below code the charachter is "@".

```python
email='Santanu123@gmail.com'
for i in email:
    if (i=='@'):
        break
    print(i, end='')
```

# Continue
#### Now dont stop the iteration comletely on specific charachter but just skip the charachter and continue the iteration untill the end.
#### Now just imagine that i want to get the domain part in the next line.

```python
email='Santanu123@gmail.com'
for i in email:
    if (i=='@'):
        print(' ')
        continue
    print(i, end='')
```
